```python
import pandas as pd
df = pd.read_csv("top-5000-youtube-channels.csv")
df.head()
```

```python
df.columns
```

```python
from textblob import TextBlob
def get_sentiment(text):
   blob = TextBlob(str(text))
   polarity = blob.sentiment.polarity
   if polarity > 0:
     return "positive"
   elif polarity < 0:
     return "negative"
   else:
     return "neutral"
```

```python
df['Sentiment'] = df['Channel name'].apply(get_sentiment)
```

```python
df[['Channel name','Grade','Sentiment']].head()
```

```python
import matplotlib.pyplot as plt
df['Sentiment'].value_counts().plot(kind='bar', color=['green','red','gray'])
plt.title("Sentiment Analysis of YouTube Channel")
plt.xlabel("Sentiment")
plt.ylabel("Count")
plt.show()
```
